```python
credits = pd.read_csv("../data/tmdb_5000_credits.csv")
movies = pd.read_csv("../data/tmdb_5000_movies.csv")
credits.head()
```

## Descripción de los datasets:

El primer dataset, credits, contiene:

- **movie_id**: Un identificador único para cada película.
- **title**: El título de cada película.
- **cast**: Los nombres de los actores principales y secundarios.
- **crew**: Los nombres del director, editor, compositor, guionista, etc.

El segundo dataset, mobies, contiene las siguientes características:

- **budget**: El presupuesto con el que se hizo la película.
- **genre**: Los géneros de la película, como Acción, Comedia, Thriller, etc.
- **homepage**: Un enlace a la página web oficial de la película.
- **id**: Este es, de hecho, el `movie_id` del primer conjunto de datos.
- **keywords**: Las palabras clave o etiquetas relacionadas con la película.
- **original_language**: El idioma en el que fue hecha la película.
- **original_title**: El título original de la película antes de su traducción o adaptación.
- **overview**: Una breve descripción de la película.
- **popularity**: Una cantidad numérica que especifica la popularidad de la película.
- **production_companies**: Las casas de producción de la película.
- **production_countries**: El país en el que se produjo la película.
- **release_date**: La fecha en la que se estrenó la película.
- **revenue**: Los ingresos generados mundialmente por la película.
- **runtime**: La duración de la película en minutos.
- **status**: El estado de la película, "Released" (Estrenada) o "Rumored" (Rumoreada).
- **tagline**: El lema de la película.
- **title**: El título de la película.
- **vote_average**: La calificación promedio que recibió la película.
- **vote_count**: El número de votos recibidos.


Tartaremos de unir estos dos datasets, pero antes realizamos los siguientes pasos:

1. **Renombrar la columna `id` de `movies` a `movie_id`** para que coincida con la columna del primer dataset.

2. **Eliminar la columna `title` de `movies`**, ya que esta columna existe en ambos datasets y queremos evitar duplicados al unir los datos.

```python
movies.rename(columns={'id': 'movie_id'}, inplace=True)
movies = movies.drop(columns=['title'])
```

```python
df = credits.merge(movies, on="movie_id")
df.head()
```

Ya tenemos el dataset combinado y listo para el análisis. Ahora podemos proceder con un análisis más profundo de las variables, visualizaciones y la preparación para el modelado posterior.
